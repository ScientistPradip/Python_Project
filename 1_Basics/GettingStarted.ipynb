**Lambda Function **
Technically the point of the lambda function is to write something in a single line

```python
mul_two = lambda x: x*2
mul_two(5)
```

```python
(lambda x: x*2)(5)
```

If I want to provide multiple arguments in the lambda function

```python
(lambda x,y : x*2 + y)(5,2)
```

How's lambda actually apply in the real world? Let's go back to that previous example that we did in our function section in where we were using this calculate salary function in order to calculate total salaryu which is based on base salary times 1 + the bonus rate.
  and now lets say with this function we have the list of salary value we want to pass into it and in our case we're going to just assume that all the bonus rate are in fact 0.1% for all of these to go through it so how could we with what we know rigth now actually implememnt it without using lanbda.
    Well we can use list comprehension and we'll define a variable that we're going to be cycling through each of these saliries in salaries list so i am going to name it to name salary and we'll loop through that using for loop

```python
salary_list = [100000, 200000, 150000, 120000, 80000, 750000]

#Calculate total_salary of base_salary *(1+ bonus_rate)
def calculate_salary(base_salary, bonus_rate=.1):

  total_salary = base_salary * (1+ bonus_rate)

  return total_salary

total_salary_list = [calculate_salary(salary) for salary in salary_list]

total_salary_list
```

Now we got that code in like so many lines. Lets shrink it down

```python
salary_list = [100000, 200000, 150000, 120000, 80000, 750000]

#Calculate total_salary of base_salary *(1+ bonus_rate)
def calculate_salary(base_salary, bonus_rate=.1):

  return base_salary * (1+ bonus_rate)

total_salary_list = [calculate_salary(salary) for salary in salary_list]

total_salary_list
```

So just the calculate_salary function above is in about one line of code. Anytime we get into writing a function with line of  code as a use case for lambda function. Remember anytime we do lambda function itself we need to define it inside a paranthesis


```python
salary_list = [100000, 200000, 150000, 120000, 80000, 750000]

#Calculate total_salary of base_salary *(1+ bonus_rate)
#def calculate_salary(base_salary, bonus_rate=.1):

  #return base_salary * (1+ bonus_rate)

total_salary_list = [(lambda x:x*1.1) (salary) for salary in salary_list]

total_salary_list
```

The above example is shown in order to display that you can write with lambda functions one single line of code and write things more succinctly however this is not completly succinct. Technically we could just put salary * 1.1 and we're going to get same results and that's what you actually should do in this case below.
I just wanted to show for demonstration purposes how you could actually use lambda inside of list comprehension

```python
total_salary_list = [salary* 1.1 for salary in salary_list]
total_salary_list
```
