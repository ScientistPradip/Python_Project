```python
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt
import statistics

#Data Loading
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

#Cleanup Data /converting it into datetime format
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
df.loc[0:10, 'salary_rate': 'salary_hour_avg'] 
```

```python
median_salary_year = df['salary_year_avg'].median()
median_salary_year
```

```python
median_salary_hour = df['salary_hour_avg'].median()
```

```python
df_filled= df
df_filled['salary_year_avg'].fillna(median_salary_year)
```

```python
df['salary_year_avg'] =df_filled['salary_year_avg'].fillna(median_salary_year)
df['salary_hour_avg'] =df_filled['salary_hour_avg'].fillna(median_salary_hour)
```

```python
df.loc[:10,'salary_year_avg':'salary_hour_avg']
```

```python
df_unique = df_filled
df_unique = df_unique.drop_duplicates()
print('Length of original df is:        ', len(df_filled))
print('Length of drop duplicates df is:', len(df_unique))
print('Rows dropped is:       ', len(df_filled) - len(df_unique))
```

```python
df_unique = df_unique.drop_duplicates(subset=['job_title', 'company_name'])
print('Length of original df is:        ', len(df_filled))
print('Length of drop duplicates df is:', len(df_unique))
print('Rows dropped is:       ', len(df_filled) - len(df_unique))
```
