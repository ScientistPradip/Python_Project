```python
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

#Data Loading
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

#Cleanup Data /converting it into datetime format
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
df_US = df[df ['job_country'] == 'United States'].copy()
df_US
```

```python
df_US['job_posted_month'] = df_US['job_posted_date'].dt.strftime('%B')
df_US
```

```python
df_US_pivot = df_US.pivot_table(index='job_posted_month', columns='job_title_short', aggfunc='size')
df_US_pivot  
```

```python
df_US_pivot.reset_index(inplace = True)
df_US_pivot['month_no'] = pd.to_datetime(df_US_pivot['job_posted_month'], format='%B').dt.month
df_US_pivot.sort_values('month_no', inplace = True)
df_US_pivot.set_index('job_posted_month', inplace=True) 
df_US_pivot.drop(columns='month_no', inplace=True)
```

```python
df_US_pivot
```

```python
df_US_pivot.plot(kind='line')
```

```python
top_job = df_US['job_title_short'].value_counts().head(3)
top_job=top_job.index.tolist() 
```

```python
df_US_pivot[top_job].plot(kind='line')
plt.title('Monthly job Postings for Top Data Jobs in US')
plt.ylabel('Job Count')
plt.xlabel('2023')
plt.legend()
plt.show()
```
