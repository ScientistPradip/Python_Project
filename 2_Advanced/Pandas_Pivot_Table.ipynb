```python
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt

#Data Loading
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

#Cleanup Data /converting it into datetime format
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
df.job_title_short.value_counts()
```

There are 4 functions for pivot table they are: values, index, columns and aggfunctions. But for now we goingt to jump into index(that used group by) and aggrigation functions (sum, min, max,etc)

To get the Count of the job title using grp and also another technique lets use another teniq 1st 

note: Size -->  counts all rows in a group (Value/Column not required)
      Count --> counts non-NA values in each group (Value/Column required)

```python
df.pivot_table(index='job_title_short', aggfunc='size')
```

instead i could just do the grp by on job title short and also run that size aggregation func and we get the same exact result and i am sticking with group by as it looks more simpler

```python
df.groupby('job_title_short').size()
```

Getting median of the every job title using values parameter from pandas.dataframe.pivot_table docs 

```python
df.pivot_table(values ='salary_year_avg', index='job_title_short', aggfunc='median')
```

same result extraction using median method again 

```python
df.groupby('job_title_short')['salary_year_avg'].median()
```

Median salary for these different job titles broken down by country (using the last one column paramenter this time) pandas.DataFrame.pivot_table

```python
df.pivot_table(values ='salary_year_avg', index='job_title_short', columns='job_country', aggfunc='median')
```

```python
df.pivot_table(values ='salary_year_avg', index='job_country', columns='job_title_short', aggfunc='median')
```

Now getting median salary by country and job title (bar graph) as the topic already mentions Pivot

```python
top_countries = df['job_country'].value_counts().head(6)
top_countries
```

I care about index only not the values bruh 

```python
top_countries = df['job_country'].value_counts().head(6).index
top_countries
```

```python
top_countries = df['job_country'].value_counts().head(6).index

df_job_country_salary = df.pivot_table(
    values='salary_year_avg',
    index= 'job_country',
    columns='job_title_short',
    aggfunc='median'
) 
df_job_country_salary = df_job_country_salary.loc[top_countries]
job_titles = ['Data Analyst', 'Data Scientist', 'Data Engineer']
df_job_country_salary = df_job_country_salary[job_titles]
df_job_country_salary.plot(kind= 'bar')
plt.ylabel('Median Salary in ($USD)')
plt.xlabel('')
plt.title('Median Salary by Country and Job Title')
plt.xticks(rotation=45, ha='right')


```
